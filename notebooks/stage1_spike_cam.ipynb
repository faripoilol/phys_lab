# Stage 1 Webcam CV Spike

This notebook visualizes the outputs generated by `track_marker_cam.py`.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
```

```python
run_dir = Path('../out/run1')
df = pd.read_csv(run_dir / 'data.csv')
df.head()
```

```python
plt.figure(figsize=(10, 4))
plt.plot(df['t_s'], df['displacement_px'])
plt.xlabel('t_s')
plt.ylabel('displacement_px')
plt.title('Displacement vs Time')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
